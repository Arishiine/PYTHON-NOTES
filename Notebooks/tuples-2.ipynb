## TUPLES

- A tuple is a built-in Python data structure used to store an ordered, immutable collection of elements.
- Core CharacteristicsOrdered: Elements maintain their defined sequence and are accessed via zero-based indexing.
- Immutable: Elements cannot be changed, added, or removed after creation.
- Heterogeneous: Can store multiple data types simultaneously (e.g., strings, integers, lists).
- Allow Duplicates: Identical values can occupy different index positions.

```python
# Empty tuple
empty_tuple = ()

# Tuple with multiple elements
fruits = ("apple", "banana", "cherry")

# Parentheses are optional (Tuple Packing)
colors = "red", "green", "blue" 
```

```python
# Incorrect: This creates a string
not_a_tuple = ("apple")  # type(not_a_tuple) is str

# Correct: Trailing comma defines the tuple
sole_tuple = ("apple",)  # type(sole_tuple) is tuple
```

```python
# Converting a list to a tuple
list_data = [1, 2, 3]
converted_tuple = tuple(list_data)  # (1, 2, 3)

# Converting a string to a tuple of characters
string_tuple = tuple("python")  # ('p', 'y', 't', 'h', 'o', 'n')
```

```python
# indexing and slicing

numbers = (10, 20, 30, 40, 50)

print(numbers[0])   # 10  (First element)
print(numbers[-1])  # 50  (Last element using negative indexing)
```

```python
#slicing

letters = ('a', 'b', 'c', 'd', 'e', 'f')

print(letters[1:4])    # ('b', 'c', 'd') - Indexes 1, 2, and 3
print(letters[:3])     # ('a', 'b', 'c') - Start to index 2
print(letters[::2])    # ('a', 'c', 'e') - Every second element
print(letters[::-1])   # ('f', 'e', 'd', 'c', 'b', 'a') - Reversed tuple
```

```python
#immutability

immutable_sample = (1, 2, 3)
# immutable_sample[0] = 99  --> Raises TypeError: 'tuple' object does not support item assignment
```

```python
#exception:mutable objects inside a tuple

mixed_tuple = (1, 2, ["apple", "banana"])

# This is valid because you are modifying the list, not the tuple structure
mixed_tuple[2][0] = "cherry" 

print(mixed_tuple)  # Output: (1, 2, ['cherry', 'banana'])
```

### Tuples Operators



```python
#joining and multplying
# Concatenation (+)
tuple1 = (1, 2)
tuple2 = (3, 4)
combined = tuple1 + tuple2  # (1, 2, 3, 4)

# Replication (*)
repeated = ("A",) * 3  # ('A', 'A', 'A')
```

```python
#membership testing
items = ("laptop", "mouse", "keyboard")
print("mouse" in items)      # True
print("screen" not in items)  # True
```

```python
data = (1, 3, 7, 8, 7, 5)

print(data.count(7))  # 2 # returns the number of times  specified value occurs
print(data.index(7))  # 2 (First occurrence index) returns the ndex of the first occurrenceof a specified value
#raises a value erroe if the item is not found
```

```python
# Concatenation (+)
tuple1 = (1, 2)
tuple2 = (3, 4)
combined = tuple1 + tuple2  # (1, 2, 3, 4)

# Replication (*)
repeated = ("A",) * 3  # ('A', 'A', 'A')
```

```python
stats = (10, 40, 20, 30)

print(len(stats))  # 4 (Total elements)
print(max(stats))  # 40 (Highest value)
print(min(stats))  # 10 (Lowest value)
print(sum(stats))  # 100 (Total sum)
```
