# Python Sets

A **set** is a built-in collection type that stores **unique** items.

- **Unique**: duplicates are dropped automatically.
- **Unordered**: items have no fixed position, so there is no indexing or slicing.
- **Fast lookups**: membership checks, adding, and removing are quick because sets use hashing internally.
- **Mutable** (you can add/remove items), but the items themselves must be **hashable** (e.g. numbers, strings, tuples; not lists or dicts).

## 1. Creating a Set
Use curly braces with comma-separated values.

```python
primes = {2, 3, 5, 7}
print(primes)
print(type(primes))
```

> **Careful:** `{}` on its own creates an empty **dictionary**, not a set. Use `set()` for an empty set.

```python
empty_dict = {}
empty_set = set()

print(type(empty_dict))
print(type(empty_set))
```

## 2. Type Casting with `set()`

`set()` converts any iterable (list, tuple, string, range...) into a set. This is the classic way to **remove duplicates**.

```python
visits = ["Nairobi", "Eldoret", "Nairobi", "Mombasa", "Eldoret"]

unique_visits = set(visits)
print(unique_visits)
print(len(visits), "->", len(unique_visits))

print(set("banana"))        # unique characters
print(set((1, 1, 2, 3, 3)))
```

## 3. Uniqueness and No Indexing

Duplicates vanish on creation. Because sets are unordered, you **cannot** read or assign by position.

```python
s = {"red", "green", "red", "blue", "green"}
print(sorted(s))    # sorted() just to get a stable display order
```

```python
try:
    print(s[0])
except TypeError as e:
    print("TypeError:", e)

try:
    s[0] = "yellow"
except TypeError as e:
    print("TypeError:", e)
```

You can't change an item *in place*, but you can still **add** or **remove** items (next section).

## 4. Mixed Data Types

A set can hold different types, as long as each item is hashable.

```python
mixed = {"text", 42, 3.14, (1, 2)}
print(mixed)
```

```python
try:
    bad = {"ok", [1, 2, 3]}     # a list is not hashable
except TypeError as e:
    print("TypeError:", e)
```

## 5. Adding and Removing Elements

| Method | Behaviour |
|---|---|
| `add(x)` | adds one item (ignored if already present) |
| `update(iterable)` | adds many items at once |
| `remove(x)` | removes `x`; raises `KeyError` if missing |
| `discard(x)` | removes `x`; does nothing if missing |
| `pop()` | removes and returns an arbitrary item |
| `clear()` | empties the set |

```python
tools = {"python", "sql"}

tools.add("excel")
print(sorted(tools))

tools.add("python")          # already there, no change
print(sorted(tools))

tools.update(["git", "docker", "sql"])
print(sorted(tools))
```

```python
tools.remove("excel")
print(sorted(tools))

tools.discard("not-in-set")  # safe, no error
print(sorted(tools))

try:
    tools.remove("not-in-set")
except KeyError as e:
    print("KeyError:", e)

removed = tools.pop()        # arbitrary item
print("popped:", removed)

tools.clear()
print(tools)
```

## 6. Set Operations

Sets mirror the maths of set theory. Each operation has a **method** and an **operator** form.

| Operation | Method | Operator | Result |
|---|---|---|---|
| Union | `a.union(b)` | `a \| b` | everything in either set |
| Intersection | `a.intersection(b)` | `a & b` | only items in both |
| Difference | `a.difference(b)` | `a - b` | items in `a` but not in `b` |
| Symmetric difference | `a.symmetric_difference(b)` | `a ^ b` | items in exactly one set |

```python
python_class = {"Amina", "Brian", "Cheri", "Dennis"}
sql_class = {"Cheri", "Dennis", "Esther", "Faith"}

print("Union       :", sorted(python_class | sql_class))
print("Intersection:", sorted(python_class & sql_class))
print("Difference  :", sorted(python_class - sql_class))
print("Sym. diff   :", sorted(python_class ^ sql_class))
```

```python
# The method versions give the same results
print(sorted(python_class.union(sql_class)))
print(sorted(python_class.intersection(sql_class)))
print(sorted(python_class.difference(sql_class)))
print(sorted(python_class.symmetric_difference(sql_class)))
```

## 7. Comparing Sets (Subset and Superset)

| Check | Meaning |
|---|---|
| `a <= b` / `a.issubset(b)` | every item of `a` is in `b` |
| `a < b` | proper subset (subset, and not equal) |
| `a >= b` / `a.issuperset(b)` | `a` contains all items of `b` |
| `a > b` | proper superset |
| `a.isdisjoint(b)` | no items in common |
| `a == b` / `a != b` | same / different items (order ignored) |

```python
small = {1, 2}
big = {1, 2, 3, 4}

print(small <= big)
print(small < big)
print(big >= small)
print(small.isdisjoint({8, 9}))
print({1, 2, 3} == {3, 2, 1})
```

## 8. Membership Tests
Use `in` / `not in`. This is much faster on a set than on a list when the collection is large.

```python
allowed = {"admin", "editor", "viewer"}

print("editor" in allowed)
print("guest" not in allowed)
```

```python
import time

big_list = list(range(2_000_000))
big_set = set(big_list)
target = 1_999_999

t0 = time.perf_counter(); target in big_list; t_list = time.perf_counter() - t0
t0 = time.perf_counter(); target in big_set;  t_set  = time.perf_counter() - t0

print(f"list lookup: {t_list:.6f}s")
print(f"set lookup : {t_set:.6f}s")
```

## 9. Iterating Over a Set

You can loop over a set, but don't rely on the order.

```python
for item in sorted({"c", "a", "b"}):
    print(item)
```

## 10. Frozen Sets

A **frozenset** is the immutable version of a set. You can't add or remove items, but you can still do unions, intersections, and differences (they return new frozensets). Because it is hashable, a frozenset can be used as a **dictionary key** or as an item inside another set.

```python
fs = frozenset(["a", "b", "c"])
print(fs)

try:
    fs.add("d")
except AttributeError as e:
    print("AttributeError:", e)

print(sorted(fs | frozenset(["c", "d"])))
```

```python
# Sets can't hold normal sets, but they can hold frozensets
group_of_groups = {frozenset({1, 2}), frozenset({3})}
print(group_of_groups)

# Frozenset as a dictionary key
labels = {frozenset({"red", "blue"}): "purple-ish"}
print(labels[frozenset({"blue", "red"})])
```

## 11. How Sets Work Internally

Sets are built on a **hash table**, the same idea behind dictionaries.

1. Python computes a **hash** (an integer) for each item.
2. The hash decides which slot (bucket) of the table the item goes into.
3. To check membership, Python hashes the item and looks directly in that slot, so there is no need to scan the whole collection.
4. If two items land in the same slot (a **collision**), Python resolves it so both can still be stored and found.

This is why set items must be **hashable**, and why membership checks are fast on average.

```python
print(hash("python") == hash("python"))   # same value -> same hash
print(hash(42), hash((1, 2)))

# Equal numbers hash equally, so 1 and 1.0 and True collapse into one item
print({1, 1.0, True})
```

## 12. Quick Recap

- Use a set when you need **uniqueness**, **fast membership checks**, or **maths-style comparisons** between groups.
- Use a list or tuple when **order or position** matters.
- `{}` is an empty dict; use `set()` for an empty set.
- Use `frozenset` when you need an immutable, hashable set.
