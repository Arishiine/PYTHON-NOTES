## Python Dictionaries.
- A dictionary is a built-in Python data structure used to store data values in key-value pairs.

Core Characteristics
- Ordered: Elements maintain their insertion order (as of Python 3.7+).
- Key-Driven: Values are retrieved using their corresponding keys, not numerical indexes.
- Unique Keys: Duplicate keys are not allowed. Assigning a value to an existing key overwrites the old value.- - - Mutable: You can add, change, or remove key-value pairs after creation.
- Keys Must Be Immutable: Keys must be of a hashable type (e.g., strings, integers, tuples). Lists and other dictionaries cannot be keys.

```python
# Empty dictionary
empty_dict = {}

# Dictionary with items
user = {
    "username": "coder123",
    "email": "user@example.com",
    "age": 25
}


mathias={
    'name':'Mathias',
    'class':'Data Science',
    'Age':50
}

print(mathias)
```

```python
# Using keyword arguments
employee = dict(name="Alice", role="Developer", ID=4502)

# Converting a list of tuple pairs into a dictionary
mapping = dict([("A", 1), ("B", 2)])  # {'A': 1, 'B': 2}
```

### accessing elements

```python
car = {"brand": "Ford", "model": "Mustang", "year": 2024}

print(car["model"])  # "Mustang"
# print(car["color"]) --> Raises KeyError if the key does not exist
```

```python
print(car.get("color"))         # None (No error raised)
print(car.get("color", "Red"))  # "Red" (Returns custom default value)
```

```python
profile = {"name": "Bob", "status": "Active"}

# Update an existing key
profile["status"] = "Inactive"

# Add a brand new key-value pair
profile["location"] = "Nairobi"
```

```python
stock = {"apples": 10, "bananas": 5}
stock.update({"bananas": 12, "oranges": 8})  # Overwrites bananas, adds oranges
```

## Removing Elements
- pop(key): Removes the specified key and returns its value. Raises an error if missing unless a fallback is provided.
- popitem(): Removes and returns the last inserted key-value pair as a tuple.
- del keyword: Deletes an item directly by key, or deletes the entire dictionary variable.
- clear(): Empties the entire dictionary contents.

```python
data = {"a": 1, "b": 2, "c": 3}

val = data.pop("b")  # val = 2, data is now {"a": 1, "c": 3}
last = data.popitem()  # last = ('c', 3), data is now {"a": 1}
del data["a"]  # data is now empty {}
```

```python
#iterating through dictionaries
inventory = {"shoes": 20, "shirts": 50, "hats": 15}
for item in inventory:
    print(item)  # Outputs: shoes, shirts, hats
for count in inventory.values():
    print(count)  # Outputs: 20, 50, 15

for item, count in inventory.items():
    print(f"{item}: {count}")  # Outputs: shoes: 20, etc.

```

```python
#dictionary comprehension
# Create a dictionary of number-to-square mappings
numbers = [1, 2, 3, 4]
squares = {x: x**2 for x in numbers}
print(squares)  # {1: 1, 2: 4, 3: 9, 4: 16}

# Filtering during comprehension
even_squares = {x: x**2 for x in numbers if x % 2 == 0}
```
