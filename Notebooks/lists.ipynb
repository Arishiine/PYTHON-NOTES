# Python Lists

A **list** is a built-in data structure that holds an ordered collection of items.

- **Ordered**: items stay in the order you put them in.
- **Mutable**: you can change, add, or remove items after creating the list.
- **Flexible**: a single list can mix data types (ints, strings, floats, even other lists).
- **Indexed**: items are reached by position, starting at index `0`.

## 1. Creating a List

There are three common ways to create a list.

### a) Square brackets `[]`

```python
numbers = [4, 8, 15]
fruits = ["mango", "papaya", "guava"]
mixed = [1, "two", 3.0, True]

print(numbers)
print(fruits)
print(mixed)
```

### b) The `list()` constructor
Pass any iterable (tuple, string, range, set...) and get a list back.

```python
from_tuple = list((10, 20, 30))
from_string = list("Kenya")
from_range = list(range(5))

print(from_tuple)
print(from_string)
print(from_range)
```

### c) Repeating elements with `*`
Multiplying a list repeats its contents.

```python
zeros = [0] * 6
pattern = ["a", "b"] * 3

print(zeros)
print(pattern)
```

## 2. How Lists Are Stored (Internal Representation)

A list does not hold the values themselves; it holds **references** to objects that live elsewhere in memory.

- Changing a *mutable* object inside the list (e.g. another list) changes the original object.
- "Changing" an *immutable* object (int, str, tuple) really means pointing the slot at a new object.
- Because of this, one list can happily hold different types.

```python
inner = [1, 2]
outer = [inner, "text", 99]

inner.append(3)       # modify the mutable object
print(outer)          # outer sees the change, since it only references inner

print(id(inner) == id(outer[0]))   # same object
```

## 3. Accessing Elements

Use an index in square brackets. Indexing is **zero-based**, and **negative indices** count from the end (`-1` is the last item).

```python
scores = [55, 62, 71, 88, 93]

print(scores[0])     # first
print(scores[2])     # third
print(scores[-1])    # last
print(scores[-2])    # second last
```

**Slicing** returns a new list from a range of positions: `list[start:stop:step]` (stop is excluded).

```python
print(scores[1:4])
print(scores[:3])
print(scores[::2])
print(scores[::-1])   # reversed copy
```

## 4. Adding Elements

| Method | What it does |
|---|---|
| `append(x)` | adds one item at the end |
| `insert(i, x)` | adds one item at position `i` |
| `extend(iterable)` | adds every item of an iterable at the end |

```python
items = ["a", "b"]

items.append("c")
print("append :", items)

items.insert(0, "start")
print("insert :", items)

items.extend(["x", "y"])
print("extend :", items)
```

> **Tip:** `append([1, 2])` adds the list as a *single* element, while `extend([1, 2])` adds `1` and `2` separately.

```python
t1 = [0]; t1.append([1, 2])
t2 = [0]; t2.extend([1, 2])
print(t1)
print(t2)
```

## 5. Updating Elements

Lists are mutable, so assign to an index (or a slice) to change contents.

```python
temps = [21, 23, 25, 27]
temps[0] = 20
print(temps)

temps[1:3] = [30, 31]   # replace a slice
print(temps)
```

## 6. Removing Elements

| Tool | Behaviour |
|---|---|
| `remove(x)` | deletes the **first** item equal to `x` (error if missing) |
| `pop()` / `pop(i)` | removes and **returns** the last item / item at index `i` |
| `del lst[i]` | deletes the item at index `i` (also works with slices) |
| `clear()` | empties the entire list |

```python
data = [5, 6, 7, 6, 8]

data.remove(6)
print("remove(6):", data)

last = data.pop()
print("pop()    :", data, "-> returned", last)

del data[0]
print("del [0]  :", data)

data.clear()
print("clear()  :", data)
```

## 7. Iterating Over a List

A `for` loop visits each element in order. Use `enumerate()` when you also need the index.

```python
cities = ["Nairobi", "Eldoret", "Mombasa"]

for city in cities:
    print(city)

print("---")
for i, city in enumerate(cities):
    print(i, city)
```

## 8. Nested Lists

A list can contain other lists. This is a handy way to represent a **matrix** or a small **table**. Chain indices to dig in: `grid[row][col]`.

```python
grid = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9],
]

print(grid[0])       # first row
print(grid[1][2])    # row 1, column 2

for row in grid:
    print(row)
```

## 9. Quick Reference: Other Useful Operations

```python
nums = [3, 1, 4, 1, 5]

print(len(nums))        # number of items
print(nums.count(1))    # occurrences of 1
print(nums.index(4))    # position of first 4
print(4 in nums)        # membership test

nums.sort()
print(nums)             # sorted in place
print(sorted(nums, reverse=True))   # new sorted list
```
