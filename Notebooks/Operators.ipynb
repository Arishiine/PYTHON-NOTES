## PYTHON OPERATORS

- There are many types of operators in python.Depending on the function to be performed.

- The most commonly used include :
    1. Assignment- these are used to assign values and variables
    2. Arithmetic - They are used to perform mathematical operations in python
    3. Comparison - They are used to compare two values
    4. Logical - Combines conditional statements
    5. Identity - They are used to compare objects

- Other types of operators include:
    1. Membership - They are used to test if a sequence is present in an object
    2. Bitwise - They are used to perform bitwise operations on values


### 1. ASSIGNMENT
- Are used with numeric values to perform mathematical operations
- They are 
  - -(subtraction),
  - +(addition),
  - *(multiplication),
  - /(division),%(modulus),
  - **(exponential),
  - //(floor division-rounds down and returns an integer)

```python
a=2
b=3

print(a-b)
print(a+b)
print(a*b)
print(a/b)
print(a%b)
print(a**b)
print(a//b)
```

### 2. Arithmetic Operators
- Are used to assing values to variables
- Most common are = ,+=,-=,*=,/=

```python
x=5
```

### 3. COMPARISON
- Are use to compare two values
- They return True/False based on the comparison
- Most commonly use are:
  * == equal
  * = not equal to
  * Greater than
  * Less than
  * Greater than or equal to
  * Less than or equal to

```python
x==5#true
x!=5#false
```

### 4. LOGICAL
- They combine conditions and return a boolean value. 
- The logical operators are:
   - and - returns True if both statements are true
   - or - returns True if one of the statements is true
   - not - returns True if the statement is false and vice versa

```python
x = 5
print(x > 0 and x < 10)
print(x > 0 or x < 10)
print(not(x > 0 and x < 10))
```

### 5. IDENTITY 
-  compares the objects, not if they are equal but if they are actually the same object with the same memory location.
- The identity operators are:
   * is - returns True if both variables are the same object.
   * is not - returns True if both variables are not the same object.

```python
x=[2,3,4]
print(2 in x)
print(5 not in x)
```
