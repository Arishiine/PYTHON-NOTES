# Python Tuples

A **tuple** is an ordered collection of items that **cannot be changed** once it has been created.

- **Ordered**: items keep their position and are reached by index (starting at `0`).
- **Immutable**: no adding, removing, or replacing items in place.
- **Heterogeneous**: one tuple can hold different data types.
- Think of it as a read-only cousin of the list.

## 1. Creating a Tuple

Place items inside parentheses `()`, separated by commas. A tuple can hold any number of items.

```python
empty = ()
coords = (36.07, -0.28)
days = ("Mon", "Tue", "Wed")

print(empty)
print(coords)
print(days)
```

### Other ways to build one

```python
# From a list
from_list = tuple([10, 20, 30])
print(from_list)

# From a string (each character becomes an item)
from_string = tuple("Eldoret")
print(from_string)

# From a range
from_range = tuple(range(1, 6))
print(from_range)

# Parentheses are optional: the commas make the tuple
no_parens = 1, 2, 3
print(no_parens, type(no_parens))
```

### The single-element gotcha

A one-item tuple needs a **trailing comma**. Without it, Python just sees parentheses around a value.

```python
not_a_tuple = (5)
single = (5,)

print(type(not_a_tuple))
print(type(single))
```

### Mixed data types

Tuples can contain ints, strings, floats, booleans, lists, dictionaries, and even other tuples.

```python
record = ("Amina", 29, 1.65, True, ["Python", "SQL"], {"city": "Nairobi"})
print(record)

for item in record:
    print(type(item).__name__, "->", item)
```

## 2. Accessing Elements

Use **indexing** (zero-based; negative indices count from the end) and **slicing** exactly like you would with a list.

```python
langs = ("Python", "R", "SQL", "Julia", "Scala")

print(langs[0])    # first
print(langs[2])    # third
print(langs[-1])   # last
print(langs[-2])   # second last
```

## 3. Slicing

Slicing builds a **new tuple** from part of an existing one: `tuple[start:stop:step]`. The `stop` index is excluded.

```python
letters = tuple("PYTHONISFUN")

print(letters[2:6])     # positions 2 to 5
print(letters[:4])      # from the start
print(letters[7:])      # to the end
print(letters[::2])     # every second item
print(letters[::-1])    # reversed
print(letters[:])       # full copy
```

> **Tip:** a negative step walks backwards, so `[::-1]` is the quickest way to reverse a tuple.

## 4. Concatenation and Repetition

- `+` joins tuples into a **new** tuple.
- `*` repeats a tuple.
- Both sides of `+` must be tuples. Mixing a tuple with a list raises a `TypeError`.

```python
a = (1, 2, 3)
b = ("x", "y")

print(a + b)
print(b * 3)
```

```python
try:
    print(a + [4, 5])
except TypeError as e:
    print("TypeError:", e)
```

## 5. Immutability in Action

Trying to change an item (or delete one) fails.

```python
t = (1, 2, 3)

try:
    t[0] = 99
except TypeError as e:
    print("TypeError:", e)

try:
    del t[1]
except TypeError as e:
    print("TypeError:", e)
```

**Important nuance:** immutability applies to the tuple's *slots*. If an item is itself a mutable object (like a list), that object can still be modified.

```python
box = (1, [10, 20], 3)
box[1].append(30)
print(box)
```

## 6. Deleting a Tuple

You cannot delete individual items, but you can delete the **whole tuple** with `del`. Using the name afterwards raises a `NameError`.

```python
temp = (0, 1, 2, 3, 4)
del temp

try:
    print(temp)
except NameError as e:
    print("NameError:", e)
```

## 7. Tuple Unpacking

Unpacking assigns each item of a tuple to its own variable. The number of variables must match the number of items.

```python
person = ("Wanjiku", "Data Scientist", "Nakuru")
name, role, town = person

print(name)
print(role)
print(town)
```

A handy use: swapping values without a temporary variable.

```python
x, y = 5, 12
x, y = y, x
print(x, y)
```

### Unpacking with the star operator `*`

Put `*` in front of one variable to make it **collect all the leftover items** into a list. This is useful when you only care about some positions.

```python
nums = (10, 20, 30, 40, 50)

first, *middle, last = nums
print(first)
print(middle)    # note: a list
print(last)
```

```python
head, *rest = nums
print(head, rest)

*start, final = nums
print(start, final)
```

## 8. Useful Tuple Methods and Functions

Because tuples can't change, they only have two methods: `count()` and `index()`. The usual built-ins work too.

```python
scores = (70, 85, 70, 92, 60)

print(scores.count(70))     # how many times 70 appears
print(scores.index(92))     # position of first 92
print(len(scores))
print(max(scores), min(scores), sum(scores))
print(sorted(scores))       # returns a new *list*
print(85 in scores)
```

## 9. Tuple vs List

| | List | Tuple |
|---|---|---|
| Syntax | `[1, 2, 3]` | `(1, 2, 3)` |
| Mutable | Yes | No |
| Methods | many (`append`, `remove`, `sort`...) | only `count`, `index` |
| Typical use | data that changes | fixed records, coordinates, return values |
| Can be a dict key | No | Yes (if all items are hashable) |

Tuples are also slightly lighter in memory and a bit faster to create.

```python
import sys

print("list :", sys.getsizeof([1, 2, 3, 4, 5]), "bytes")
print("tuple:", sys.getsizeof((1, 2, 3, 4, 5)), "bytes")

# Tuples as dictionary keys
locations = {(36.07, -0.28): "Nakuru"}
print(locations[(36.07, -0.28)])
```
