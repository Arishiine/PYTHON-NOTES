# DATA TYPES AND VARIABLES

## DATA TYPES
- These are ways to classify data items
- Represent the kind of value which determines what operations can be performed on that data.
- The following are inbuilt data types in python:
    - Text type - str
    - Numeric Type - int,float,complex
    - Sequence type - list,tuple,range
    - Mapping type - dict
    - Set type - set, frozenset
    - Boolean Type -bool
    - Binary type - bytes,bytearray,memoryview
    - None type - NoneType

```python
# getting the data type
#use the type funtion-- type()
print(type('name'))
v=True
print(type(v)) #getting the data type of a variable
```

```python
#set data type
#assigning a value to a variable
m=bool(2)
print(m)
```

#### TEXT/STRING DATA TYPE
- Always surrounded by quotation marks,either double "" or single''
- For a long string,use 3 double quotes
- sequence of characters written inside a quote

```python
##comment in python
"This is a string in python,It should be in quotes"
'Hello World'
```

```python
#loop through a string
for w in "orange":
    print(w)
```

```python
#getting the lenghth of a string
#use len() function
print(len('Hello World')) #10 characters + 1 space =11
```

```python
#check if phrase is present in string
print('a' in 'watermelon') #output retuens a boolean
#use not - output will also be boolean
print('v' in "orange")
```

```python
#string arrays
name="Yellow"
print(name[1]) #indexing starts from 0
print(name[-6]) #negative indexing starts with the furthest character
#accessing an index out of range will bring an index error
```
