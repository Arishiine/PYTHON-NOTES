# DATA TYPES AND VARIABLES

## VARIABLES
- A variable is a container used to store values
- Are created the exact moment you assign a value to them using the assignment operator (=) i.e **variable = value**
- Rules for naming:
    - Must start with a letter or an underscore (_).
    - Cannot start with a number.Can only contain alphanumeric characters and underscores (A-z, 0-9, and _).
    - Case-sensitive: age, Age, and AGE are three different variables.
    - Cannot use Python's built-in reserved keywords like if, for, class, or import
- Naming Best Practices
    - Use snake_case for multi-word variables (e.g., user_registration_date).
    - Give descriptive names instead of vague ones (e.g., use total_price instead of tp)

```python
#creating a variable-variable_name=value
name='Data Analysis'
#here name is the variable and 'Data Analysis' is the value assigned to it
```

## DATA TYPES
- These are ways to classify data items
- Represent the kind of value which determines what operations can be performed on that data.
- The following are inbuilt data types in python:
    - Text type - str
    - Numeric Type - int,float,complex
    - Sequence type - list,tuple,range
    - Mapping type - dict
    - Set type - set, frozenset
    - Boolean Type -bool
    - Binary type - bytes,bytearray,memoryview
    - None type - NoneType

```python
# getting the data type
#use the type funtion-- type()
print(type('name'))
v=True
print(type(v)) #getting the data type of a variable
```

```python
#set data type
#assigning a value to a variable
m=bool(2)
print(m)
```
