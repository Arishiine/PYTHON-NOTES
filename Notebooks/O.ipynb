- In Python, the two fundamental functions for handling interaction are
  -  input() for receiving data from a user and 
  -   print() for displaying data to the screen

1. Standard Input: input()The input() function pauses program execution and waits for the user to type something and press Enter.
 - Default Type: It always reads data as a string (str), even if the user types a number.**Prompt: You can pass an optional message inside the parentheses to guide the user**

```python
# Taking a text input
name = input("Enter your name: ")
print("Hello,", name)
```

**Converting Input Types (Type Casting)Because input() returns a string, you must explicitly convert it using functions like int() or float() if you want to perform math operations.**


```python
#Convert to integer
age = int(input("Enter your age: "))

# Convert to float
price = float(input("Enter product price: "))
```
