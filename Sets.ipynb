## SETS

* A set is a built-in Python data structure used to store an unordered collection of unique elements.
- Core CharacteristicsUnordered:
   1. Elements have no defined sequence. They cannot be accessed via indexing or slicing.
   2. Unique Elements: Duplicate values are automatically discarded.
   3. Mutable: You can add or remove items from the set itself.
   4. Unchangeable Elements: While the set is mutable, the individual elements inside it must be of an immutable (hashable) data type (e.g., strings, integers, tuples). You cannot put a list or another set inside a set.

```python
#creating a set
# Set with multiple elements
fruits = {"apple", "banana", "cherry"}

# Duplicates are automatically removed
numbers = {1, 2, 2, 3, 3, 3}
print(numbers)  # {1, 2, 3}
```

```python
#empty set
# Incorrect: This creates a dictionary
not_a_set = {}  # type(not_a_set) is dict

# Correct: Defines an empty set
true_set = set()  # type(true_set) is set
```

```python
#constructing a set
# Converting a list to a set (excellent for removing duplicates)
duplicate_list = [1, 2, 2, 3, 4, 4]
unique_set = set(duplicate_list)  # {1, 2, 3, 4}

# Converting a string to a set of unique characters
char_set = set("letters")  # {'l', 'e', 't', 'r', 's'} (order varies)
```

```python
#adding elements
colors = {"red", "green"}

colors.add("blue")  # {"red", "green", "blue"}
colors.update(["yellow", "orange"])  # {"red", "green", "blue", "yellow", "orange"}
```

```python
#removing elements
items = {"laptop", "mouse", "keyboard"}

items.remove("mouse")  # {"laptop", "keyboard"}
items.discard("screen")  # Safe operation: does nothing, no error raised
```

- Mathematical Set Operations

* Unions

- UnionCombines elements from both sets (duplicates excluded).
Operator: set_a | set_bMethod: set_a.union(set_b)Result: {1, 2, 3, 4, 5, 6}
